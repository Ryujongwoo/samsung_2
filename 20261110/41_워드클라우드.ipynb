generate() 메소드는 인수로 지정한 텍스트에서 가장 많이 출현한 단어를 100%로 설정하고 가장 많이 출현한 단어와의 비율을 리턴한다.  
words_ 속성으로 단어별 출현 빈도수 비율을 얻어온다.

```python
plt.figure(figsize=[8, 6])
# imshow() 메소드는 워드클라우드 객체를 이미지로 출력한다.
plt.imshow(wc, interpolation='bilinear')
# axis() 메소드로 워드클라우드에 나타는 눈금을 표시하지 않는다.
plt.axis('off')
plt.show()
```

WordCloud 클래스의 생성자로 max_words를 넘겨주면 워드클라우드에 표시될 단어의 개수를 지정할 수 있다.

```python
wc = WordCloud(max_font_size=50, max_words=50).generate(text)
plt.figure(figsize=[8, 6])
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.show()
```

## 워드클라우드에 그레이스케일을 적용할 함수를 선언한다.

word: 워드클라우드로 표시할 단어  
font_size: 워드클라우드로 표시할 단어의 크기  
position: 워드클라우드로 표시할 단어의 위치  
orientation: 워드클라우드로 표시할 단어의 방향  
random_state=None: 적당한 숫자를 넘겨주면 워드클라우드로 표시되는 단어들의 위치가 고정된다.  
**args: 워드클라우드로 표시할 단어들의 글꼴이 저장된 경로

```python
def gray_scale(word, font_size, position, orientation, random_state=None, **args):
    # print(word, font_size, position, orientation, random_state, args)
    return 'hsl(0, 0%, {}%)'.format(random.randint(0, 60))
```

워드클라우드에 그레이스케일을 적용하는 함수는 recolor() 메소드의 color_func으로 지정한다.

```python
plt.figure(figsize=[9, 9])
plt.imshow(
    wc.recolor(color_func=gray_scale, random_state=200), 
    interpolation='bilinear')
plt.axis('off')
plt.show()
```

# 한글 워드클라우드

```python
from konlpy.corpus import kolaw
from konlpy.tag import Okt
okt = Okt()
```

```python
kolaw_corpus = kolaw.open('constitution.txt').read()
print(kolaw_corpus[:40])
```

## 대한민국헌법 텍스트에서 형태소 분석을 통해 명사만 추출한다.

```python
tokens = okt.nouns(kolaw_corpus)
print(len(tokens))
tokens
```

## 워드클라우드로 구현하려는 단어들의 출현 빈도수를 센다.

워드클라우드로 구현할 단어들의 빈도수를 세기위해 nltk 라이브러리를 설치하고 import 한다.  
`!pip install nltk`

```python
import nltk
```

Text() 클래스로 형태소 분석 결과를 넘겨 워드클라우드로 구현할 단어들의 출현 빈도수를 세서 출현 빈도수가 많은 단어부터 (단어, 개수) 형태의 튜플을 만든다.

```python
ko = nltk.Text(tokens)
print(len(ko))
print(len(ko.vocab().most_common()))
ko.vocab().most_common()
```

## 불용어를 처리한다.

형태소 분석 결과에서 1음절 단어를 모두 제거한 리스트를 만든다.

```python
tokens = [token for token in tokens if len(token) > 1]
print(len(tokens))
tokens
```

불용어 리스트를 만들고 형태소 분석 결과에서 1음절 단어를 모두 제거한 리스트에서 불용어를 제거한 리스트를 만든다.

```python
stopwords = ['기타', '사항', '경우', '이상', '거나', '다만', '대하', '누구']
tokens = [token for token in tokens if token not in stopwords]
print(len(tokens))
tokens
```

```python
ko = nltk.Text(tokens)
print(len(ko))
print(len(ko.vocab().most_common()))
ko.vocab().most_common()
```

## 워드클라우드를 구현하기 위해서 단어를 key로 하고 출현 빈도수를 value로 하는 딕셔너리를 만든다.

```python
words = dict(ko.vocab().most_common(250))
print(len(words))
print(words)
```
