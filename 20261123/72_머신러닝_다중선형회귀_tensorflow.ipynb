```python
import warnings
warnings.filterwarnings(action='ignore')
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
# [공부시간, 과외시간, 시험점수]
data = [[2, 0, 81], [4, 4, 93], [6, 2, 91], [8, 3, 97]]
xData1 = [d[0] for d in data]
xData2 = [d[1] for d in data]
yData = [d[2] for d in data]
print(f'공부 시간: {xData1}, 과외 시간: {xData2}, 공부 시간과 과외 시간에 따른 시험 점수: {yData}')
```

```python
a1 = tf.Variable(tf.random_uniform([1], 0, 10)) # 공부 시간 가중치
a2 = tf.Variable(tf.random_uniform([1], 0, 10)) # 과외 시간 가중치
b = tf.Variable(tf.random_uniform([1], 0, 100)) # 바이어스

sess = tf.Session()
sess.run(tf.global_variables_initializer())
print(f'a1 = {sess.run(a1)}, a2 = {sess.run(a2)}, b = {sess.run(b)}')
```

```python
Y = a1 * xData1 + a2 * xData2 + b
```

```python
rmse = tf.sqrt(tf.reduce_mean(tf.square(Y - yData)))
gradient_descent = tf.train.GradientDescentOptimizer(0.1).minimize(rmse)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())

for i in range(3001):
    sess.run(gradient_descent)
    if i % 200 == 0:
        print('epoch: {:4d}, rmse: {:9.6f}, a1: {:9.6f}, a2: {:9.6f}, b: {:9.6f}'.format(i, sess.run(rmse), sess.run(a1)[0], sess.run(a2)[0], sess.run(b)[0]))
```
