```python
import warnings
warnings.filterwarnings(action='ignore')
import tensorflow.compat.v1 as tf # tensorflow 1.x 버전을 사용하겠다고 선언한다.
tf.disable_v2_behavior() # tensorflow 2.x 버전의 문법을 사용하기 않겠다고 선언한다.
print(tf.__version__)
```

## 상수, 변수, 수식을 만든다. => 세션에 넣는다. 변수는 반드시 초기화 시킨다. => 실행한다.

tensorflow 상수

constant() 메소드로 tensorflow에서 사용할 상수를 정의한다.

```python
a = tf.constant(100) # 상수 정의
print(f'a = {a}')
b = tf.constant(200)
print(f'b = {b}')

# tensorflow 1.x 버전은 반드시 세션이라는 작업 영역을 만든 후 상수, 변수, 수식을 대입한 후 실행한다.
# Session 클래스로 tensorflow 1.x 버전에서 사용할 세션을 정의한다.
sess = tf.Session()

# run() 메소드로 세션의 내용을 실행한다.
print(f'a = {sess.run(a)}')
print(f'b = {sess.run(b)}')
```

tensorflow 변수

Variable 클래스로 tensorflow에서 사용할 변수 객체를 정의한다.

```python
c = tf.Variable(300) # 변수 정의
print(f'c = {c}')
d = tf.Variable(400)
print(f'd = {d}')

sess = tf.Session()

# tensorflow 변수 초기화
# Variable 클래스로 생성한 tensorflow 변수는 사용하기 전에 반드시 초기화시켜야 한다.
# global_variables_initializer() 메소드로 tensorflow 변수를 초기화시킨다.
init = tf.global_variables_initializer()
sess.run(init)

print(f'c = {sess.run(c)}')
print(f'd = {sess.run(d)}')

# tensorflow 변수 초기화를 실행한 후 정의한 변수는 다시 초기화시켜야 한다.
e = tf.Variable(400)
print(f'e = {e}')
# print(f'e = {sess.run(e)}')

sess.run(tf.global_variables_initializer())
print(f'e = {sess.run(e)}')
```

tensorflow 변수

```python
a = tf.constant(100)
b = tf.constant(200)
c = tf.Variable(300)
d = tf.Variable(400)

e = a + b
print(f'e = {e}')
f = tf.add(a, b)
print(f'f = {f}')

g = c * d
print(f'g = {g}')
h = tf.multiply(c, d)
print(f'h = {h}')

sess = tf.Session()
sess.run(tf.global_variables_initializer())

print(f'e = {sess.run(e)}')
print(f'f = {sess.run(f)}')
print(f'g = {sess.run(g)}')
print(f'h = {sess.run(h)}')
```
