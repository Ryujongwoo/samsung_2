```python
import warnings
warnings.filterwarnings(action='ignore')
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

## placeholder는 tensorflow 1.x 버전에서 사용할 여러건의 데이터를 저장하는 배열이다.

placeholder(dtype[, shape, name])  
`dtype`: placeholder에 저장할 데이터의 자료형을 지정한다.  
`shape`: placeholder에 저장할 데이터의 차원을 지정한다. 생략가능  
name: placeholder의 이름을 지정한다. 생략가능

```python
# placeholder에 저장할 데이터를 선언한다.
xData = [1, 2, 3, 4, 5]
yData = [5, 4, 3, 2, 1]

# placeholder() 메소드로 tensorflow 1.x 버전에서 사용할 placeholder를 선언한다.
x = tf.placeholder(dtype=tf.int32) # xData의 데이터를 저장할 placeholder를 선언한다.
y = tf.placeholder(dtype=tf.int32) # yData의 데이터를 저장할 placeholder를 선언한다.

print(f'x = {x}')
print(f'y = {y}')

sess = tf.Session()
sess.run(tf.global_variables_initializer())

# placeholder를 사용하는 수식을 만든다.
z = x + y
# 아래의 수식을 실행하면 x, y placeholder에 데이터가 채워지지 않아서 에러각 발생된다.
# print(f'z = {sess.run(z)}')

# placeholder를 사용하는 수식은 반드시 수식에 사용할 데이터를 placeholder에 대입한 후 실행해야 한다.
# placeholder에 데이터를 대입하는 방법은 feed_dict 인자에 딕셔너리 형태로 {placeholder: 데이터, placeholder: 데이터, ...}와 같은 방식을 사용한다.
result = sess.run(z, feed_dict={x: xData, y: yData})
print(result)
```
