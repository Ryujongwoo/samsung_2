```python
import warnings
warnings.filterwarnings(action='ignore')
from bs4 import BeautifulSoup
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
```

가상 크롬을 띄워서 전체 인기 댓글을 크롤링한다.

```python
option = Options()
option.add_argument('--window-size=1800,1000')
driver = webdriver.Chrome(options=option)
driver.get('https://comic.naver.com/webtoon/detail?titleId=812354&no=274&week=thu')
driver.implicitly_wait(5)

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
reviews = soup.select('p._text_8n8z9_16')
# print(len(reviews))
```

전체 인기 댓글의 개수만큼 반복하며 각 인기 댓글의 답글들을 크롤링한다.

```python
for i in range(len(reviews)):
    print('{:2d}번째 인기 댓글: {}'.format(i + 1, reviews[i].text))
    print('=' * 100)
    
    # 답글을 크롤링할 인기 댓글의 '답글' 버튼을 찾아서 클릭한다.
    xpath = '//*[@id="wcc_root"]/section/ul/li[' + str(i + 1) + ']/div[2]/div[1]/button'
    # print(xpath)
    driver.find_element(By.XPATH, xpath).click()
    time.sleep(1)
    
    # 답글 버튼을 클릭하면 나타나는 답글들을 크롤링한다.
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    replys = soup.select('div.Ug1Ok p._text_8n8z9_16')
    
    for index, reply in enumerate(replys, start=1):
        print('{:2d}번째 인기 댓글의 {:2d}번째 답글 {}'.format(i + 1, index, ' '.join(replys[index - 1].text.split('\n'))))
    print('이상 {:2d}번째 인기 댓글의 답글'.format(i + 1))
    print('=' * 100)
    
    # 다음 인기 댓글의 답변을 크롤링해야 하므로 '답글 접기' 버튼을 찾아서 클릭한다.
    driver.find_element(By.XPATH, '//*[@id="wcc_root"]/section/ul/li[' + str(i + 1) + ']/div[3]/button').click()
    time.sleep(1)
```
