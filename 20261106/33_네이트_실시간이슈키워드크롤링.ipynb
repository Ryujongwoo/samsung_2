```python
import warnings
warnings.filterwarnings(action='ignore')
import requests
```

# 네이트 실시간 이슈 키워드 크롤링

```python
targetSite = 'https://www.nate.com/js/data/jsonLiveKeywordDataV1.js?v=202611060900'
response = requests.get(targetSite)
print(response)
```

```python
issues = response.text
print(type(response.text))
print(issues)
```

```python
ranks = response.json()
print(type(ranks))
print(ranks)
```

```python
for rank in ranks:
    print(f'{rank[0]:>2s}위 {rank[1]}', end=' ')
    if rank[2] == '+':
        print(f'[▲{rank[3]}]')
    elif rank[2] == '-':
        print(f'[▼{rank[3]}]')
    elif rank[2] == 'n':
        print(f'[new]')
    else:
        print(f'[-]')
```
