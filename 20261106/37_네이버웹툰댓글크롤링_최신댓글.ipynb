```python
import warnings
warnings.filterwarnings(action='ignore')
from bs4 import BeautifulSoup
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
```

가상 크롬을 띄워서 전체 최신 댓글을 크롤링한다.

```python
option = Options()
option.add_argument('--window-size=1800,1000')
driver = webdriver.Chrome(options=option)
driver.get('https://comic.naver.com/webtoon/detail?titleId=812354&no=274&week=thu')
driver.implicitly_wait(5)

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
```

## 최신 댓글 버튼을 찾아 클릭한다.

```python
driver.find_element(By.XPATH, '//*[@id="radix-_r_c_-trigger-2"]').click() # 찐팬 댓글 XPATH: //*[@id="radix-_r_c_-trigger-3"]
time.sleep(1)
```

## 더보기 버튼을 찾아서 더보기 버튼이 없을때까지 클릭한다.

```python
while True:
    try:
        driver.find_element(By.XPATH, '//*[@id="wcc_root"]/section/button').click()
        time.sleep(1)
    except:
        break
```

## 최신 댓글 전체를 읽는다.

```python
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
reviews = soup.select('p._text_8n8z9_16')
print(len(reviews))
```

```python
for index, review in enumerate(reviews, start=1):
    print('{:3d}번째 댓글: {}'.format(index, ' '.join(review.text.split('\n'))))
```

```python
with open('./output/reviews.txt', 'w', encoding='utf-8') as file:
    for index, review in enumerate(reviews, start=1):
        file.write('{}\n'.format(' '.join(review.text.split('\n'))))
print('reviews.txt 파일에 저장완료')
```
