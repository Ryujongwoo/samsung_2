```python
import warnings
warnings.filterwarnings(action='ignore')
from bs4 import BeautifulSoup
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
```

selenium 3.x 버전까지는 `=>` 왼쪽의 메소드를 사용했고 selenium 4.x 버전 부터는 `=>` 오른쪽의 메소드를 사용한다.

find_element_by_class_name('') => find_element(`By`.CLASS_NAME, '')  
find_element_by_id('') => find_element(`By`.ID, '')  
find_element_by_name('') => find_element(`By`.NAME, '')  
find_element_by_tag_name('') => find_element(`By`.TAG_NAME, '')  
find_element_by_css_selector('') => find_element(`By`.CSS_SELECTOR, '')  
find_element_by_xpath('') => find_element(`By`.XPATH, '')  

selenium 4.x 버전의 `By`를 사용하기 위해서 By를 import 한다.

```python
from selenium.webdriver.common.by import By
```

```python
option = Options()
option.add_argument('--window-size=1800,1000')
driver = webdriver.Chrome(options=option)
driver.get('https://comic.naver.com/webtoon/detail?titleId=812354&no=274&week=thu')
driver.implicitly_wait(5)

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
reviews = soup.select('p._text_8n8z9_16')
# print(len(reviews))

print('{:2d}. {}'.format(3, reviews[2].text))
```

## 인기댓글의 답글 버튼을 찾아 클릭한다.

```python
driver.find_element(By.XPATH, '//*[@id="wcc_root"]/section/ul/li[3]/div[2]/div[1]/button').click()
time.sleep(1)
```

## 버튼을 찾아 클릭하면 화면이 변경되므로 페이지를 다시 읽어온다.

```python
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')

replys = soup.select('div.Ug1Ok p._text_8n8z9_16')
# print(len(replys))

for index, reply in enumerate(replys, start=1):
    print(f'{index:2d}. {reply.text}')
```
