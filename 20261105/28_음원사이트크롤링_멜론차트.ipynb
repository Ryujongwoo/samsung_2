```python
import warnings
warnings.filterwarnings(action='ignore')
import requests
from bs4 import BeautifulSoup
from datetime import datetime as dt
```

```python
targetSite = 'https://www.melon.com/chart/index.htm'
headers = {'User-agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64; Trident/7.0; MSAppHost/2.0; rv:11.0) like Gecko'}
reponse = requests.get(targetSite, headers=headers)
# print(reponse)
html = reponse.text
# print(html)
soup = BeautifulSoup(html, 'html.parser')
# print(soup)
```

노래 제목 크롤링

```python
titles = soup.findAll('div', {'class': 'ellipsis rank01'})
print(len(titles))
print(titles)
```

```python
for index, title in enumerate(titles, start=1):
    print('{}위 {}'.format(index, title.text.strip().split('\n')[0]))
```

아티스트 크롤링

```python
artists = soup.findAll('span', {'class': 'checkEllipsis'})
print(len(artists))
print(artists)
```

```python
for index, artist in enumerate(artists, start=1):
    print('{}위 {}'.format(index, artist.text.strip().split(',')[0]))
```

```python
print(dt.now())
for i in range(len(artists)):
    artist = artists[i].text.strip().split(',')[0]
    title = titles[i].text.strip().split('\n')[0]
    print('{:3d}위 {} - {}'.format(i + 1, artist, title))
```

```python
with open('./output/melonTOP100.txt', 'w', encoding='utf-8') as file:
    file.write(str(dt.now()) + '\n')
    for i in range(len(artists)):
        artist = artists[i].text.strip().split(',')[0]
        title = titles[i].text.strip().split('\n')[0]
        file.write('{:3d}위 {} - {}\n'.format(i + 1, artist, title))  
print('melonTOP100.txt 파일에 쓰기 완료')
```
