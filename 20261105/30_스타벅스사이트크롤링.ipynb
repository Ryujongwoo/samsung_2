# 시도 코드 크롤링

Request URL을 이용해서 스타벅스 서버로 시도 코드를 요청한다.

```python
targetSite = 'https://www.starbucks.co.kr/store/getSidoList.do'
# Request URL(https://www.starbucks.co.kr/store/getSidoList.do)을 요청하는 method가 POST 방식 이므로 requests 모듈의 post() 메소드를 사용해서 서버에 요청한다.
# POST 방식으로 요청할 때는 post() 메소드의 data 인자에 딕셔너리 형태로 서버로 전송할 데이터를 넣어줘야 한다. 서버로 전송할 데이터가 없는 경우에는 생략한다.
response = requests.post(url=targetSite, data={
    # 'rndCod': 'S5PFCRA78N' # rndCod는 서버로 전달하지 않아도 상관없다.
})
# print(response)
# 응답받은 정보 중에서 json 형태의 문자열만 얻어온다.
print(type(response.text))
print(response.text)
```

응답받은 문자열을 json 모듈의 loads() 메소드나 requests 모듈로 요청한 결과가 저장된 변수(response)에서 json() 메소드를 사용해서 응답받은 json 형태의 문자열을 파이썬에 사용할 수 있는 자료형인 `{}가 있으면 딕셔너리로 {}가 없으면 리스트`로 변환한다.

json 모듈의 loads() 메소드를 사용하는 경우

```python
sidoList = json.loads(response.text)
print(type(sidoList))
print(sidoList.keys())
```

requests 모듈로 요청한 결과가 저장된 변수에서 json() 메소드를 사용하는 경우

```python
sidoList = response.json()
print(type(sidoList))
print(sidoList.keys())
```

딕셔너리 형태로 변환된 데이터에서 시도 코드 데이터가 할당된 데이터를 얻어온다.

```python
print(len(sidoList['list']))
print(sidoList['list'])
```

시도 코드 데이터의 개수만큼 반복하며 시도코드(sido_cd)를 key로 하고 시도이름(sido_nm)를 value로 하는 딕셔너리를 만든다.

```python
sido_list = {}
for sido in sidoList['list']:
    # print('{}: {}'.format(sido['sido_cd'], sido['sido_nm']))
    sido_list[sido['sido_cd']] = sido['sido_nm']
print(sido_list)
```

# 시도 코드 별 구군 코드 크롤링

```python
sido_cd = input('구군 코드를 검색하려는 시도 코드를 입력하세요: ')
targetSite = 'https://www.starbucks.co.kr/store/getGugunList.do'
response = requests.post(url=targetSite, data={
    'sido_cd': sido_cd
})
# print(response)
# print(response.text)

gugunList = response.json()
# print(type(gugunList))
# print(type(gugunList['list']))
# print(len(gugunList['list']))
# print(gugunList)

# 구군 데이터 개수만큼 반복하며 구군코드(gugun_cd)를 key로 하고 구군이름(gugun_nm)을 value로 하는 딕셔너리를 만든다.
gugun_list = {}
for gugun in gugunList['list']:
    gugun_list[gugun['gugun_cd']] = gugun['gugun_nm']
print(gugun_list)
```
