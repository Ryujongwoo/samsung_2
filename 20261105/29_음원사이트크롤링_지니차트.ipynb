```python
import warnings
warnings.filterwarnings(action='ignore')
import requests
from bs4 import BeautifulSoup
from datetime import datetime as dt
```

```python
artists = soup.findAll('a', {'class': 'artist ellipsis'})
print(len(artists))
print(artists)
```

## 4개의 페이지를 크롤링한 결과를 하나의 리스트로 모아준다.

```python
import time
```

```python
genie = []
url = 'https://www.genie.co.kr/chart/top200?ditc=D&ymd=20261105&hh=11&rtm=Y&pg='
headers = {'User-agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64; Trident/7.0; MSAppHost/2.0; rv:11.0) like Gecko'}

for i in range(1, 5):
    targetSite = f'{url}{i}'
    reponse = requests.get(targetSite, headers=headers)
    time.sleep(1)
    html = reponse.text
    soup = BeautifulSoup(html, 'html.parser')
    
    titles = soup.findAll('a', {'class': 'title ellipsis'})
    artists = soup.findAll('a', {'class': 'artist ellipsis'})
    
    for j in range(len(titles)):
        genie.append(
            [artists[j + 5].text.strip().split('\n')[0], titles[j].text.strip().split('\n')[0]]
        )
    # ===== for j
# ===== for i
print(len(genie))
```

```python
print(genie)
```

```python
print(dt.now())
for index, item in enumerate(genie, start=1):
    print('{:3d}위 {} - {}'.format(index, item[0], item[1]))
```

```python
with open('./output/genieTOP200.txt', 'w', encoding='utf-8') as file:
    file.write(str(dt.now()) + '\n')
    for index, item in enumerate(genie, start=1):
        file.write('{:3d}위 {} - {}\n'.format(index, item[0], item[1]))  
print('genieTOP200.txt 파일에 쓰기 완료')
```
