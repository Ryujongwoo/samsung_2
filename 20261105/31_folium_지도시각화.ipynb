지도 시각화를 하기 위해서 folium을 import 한다.

```python
import folium
```

folium 라이브러리의 Map 클래스의 생성자로 location과 zoom_start를 지정해서 지도를 만든다.  
location은 위도와 경도 순서로 리스트를 지정해서 지도의 중심을 설정하고 zoom_start는 지도의 배율을 설정한다.

```python
# folium 라이브러리를 사용해서 지도를 만든다.
foliumMap = folium.Map(location=[37.442561275669, 127.136588869764], zoom_start=14)

# folium 라이브러리의 Marker 클래스의 생성자의 location으로 마커를 만들고 add_to() 메소드로 지도에 추가한다.
# popup으로 마커를 클릭했을 때 표시할 문자열을 추가할 수 있다.
# folium.Marker(location=[37.442561275669, 127.136588869764], popup='성남태평').add_to(foliumMap)

# 마커를 클릭했을 때 표시되는 문자열이 나타나는 말풍선 크기를 변경하려면 folium 라이브러리의 Popup 클래스로 말풍선의 최대 가로 크기를 지정할 수 있다.
# folium 라이브러리의 Icon 클래스를 이용해서 아이콘 색상고 모양을 지정할 수 있다.
# 아이콘 색상 목록
# red, blue, green, purple, orange, darkred, darkblue, darkgreen, cadetblue, lightred, lightblue, lightgreen, darkpurple, pink, beige, black, gray, 
# lightgray, white
# 아이콘 종류는 https://getbootstrap.com/docs/3.4/components/에서 제공하는 아이콘을 표시할 수 있다.

icon = folium.Icon(color='purple', icon='pushpin')
popup = folium.Popup('스타벅스 성남태평점', max_width=300)
folium.Marker(location=[37.442561275669, 127.136588869764], popup=popup, icon=icon).add_to(foliumMap)

icon = folium.Icon(color='darkgreen', icon='flash')
popup = folium.Popup('스타벅스 가천대학교점', max_width=300)
folium.Marker(location=[37.44941, 127.1272], popup=popup, icon=icon).add_to(foliumMap)

# save() 메소드로 folium 라이브러리를 사용해서 만든 지도를 html 파일로 저장할 수 있다.
foliumMap.save('./output/foliumMap.html')
# 지도를 jupyter nbclassic에 출력한다.
foliumMap
```
