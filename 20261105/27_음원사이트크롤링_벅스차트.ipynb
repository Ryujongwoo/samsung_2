크롤링할 사이트에 접속해서 웹 문서를 읽어오기 위해서 requests를 import 한다.

```python
import requests
```

requests를 사용해서 얻어온 웹 문서를 분석(파싱)할 수 있게 하기 위해서 BeautifulSoup을 import 한다.

```python
from bs4 import BeautifulSoup
```

```python
targetSite = 'https://music.bugs.co.kr/chart'
headers = {'User-agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64; Trident/7.0; MSAppHost/2.0; rv:11.0) like Gecko'}
```

requests 모듈의 get() 메소드로 타겟 사이트의 웹 문서를 읽어온다.

```python
reponse = requests.get(targetSite, headers=headers)
print(reponse)
# <Response [200]>: 서버가 요청을 제대로 처리했다는 뜻으로 서버가 요청한 페이지를 제공했다는 의미이다.
```

타겟 사이트에서 읽어온 정보 중에서 text로된 html 문서만 얻어온다.

```python
html = reponse.text
print(type(html))
print(html)
```

BeautifulSoup을 이용해서 얻어온 html 문서(문자열)를 html parser를 사용해서 분석(파싱)할 수 있도록 객체를 만든다.

```python
soup = BeautifulSoup(html, 'html.parser')
print(type(soup))
print(soup)
```

## html 태그와 html 태그에 지정된 속성을 이용한 크롤링

`findAll('태그이름', {'속성이름': '속성값', ...})`

노래 제목 크롤링

```python
titles = soup.findAll('p', {'class': 'title'})
print(len(titles))
print(titles)
```

```python
for index, title in enumerate(titles, start=1):
    # print(title.text.strip())
    # print(title.text.split('\n')[1])
    print('{}위 {}'.format(index, title.text.split('\n')[1]))
```

아티스트 크롤링

```python
artists = soup.findAll('p', {'class': 'artist'})
print(len(artists))
print(artists)
```

```python
for index, artist in enumerate(artists, start=1):
    # print(artist.text.strip())
    # print(artist.text.split('\n'))
    # print(artist.text.strip().split('\n')[0])
    print('{}위 {}'.format(index, artist.text.strip().split('\n')[0]))
```

컴퓨터 시스템의 날짜와 시간을 얻어오기 위해서 datetime을 import 하고 dt라는 별명을 지정한다.

```python
from datetime import datetime as dt
```

```python
print(dt.now())
for index, (artist, title) in enumerate(zip(artists, titles), start=1):
    print('{:2d}위 {} - {}'.format(index, artist.text.strip().split('\n')[0], title.text.split('\n')[1]))
```

```python
print(dt.now())
for i in range(len(artists)):
    artist = artists[i].text.strip().split('\n')[0]
    title = titles[i].text.split('\n')[1]
    print('{:2d}위 {} - {}'.format(i + 1, artist, title))
```

## 크롤링한 결과를 텍스트 파일에 저장한다.

```python
file = open('./output/bugsTOP100.txt', 'w', encoding='utf-8')
file.write(str(dt.now()) + '\n')
for i in range(len(artists)):
    artist = artists[i].text.strip().split('\n')[0]
    title = titles[i].text.split('\n')[1]
    file.write('{:3d}위 {} - {}\n'.format(i + 1, artist, title))
file.close()
print('bugsTOP100.txt 파일에 쓰기 완료')
```

```python
with open('./output/bugsTOP100.txt', 'w', encoding='utf-8') as file:
    file.write(str(dt.now()) + '\n')
    for i in range(len(artists)):
        artist = artists[i].text.strip().split('\n')[0]
        title = titles[i].text.split('\n')[1]
        file.write('{:3d}위 {} - {}\n'.format(i + 1, artist, title))    
print('bugsTOP100.txt 파일에 쓰기 완료')
```

## 크롤링 결과가 저장된 텍스트 파일을 읽어서 화면에 출력한다.

```python
try:
    file = open('./output/bugsTOP100.txt', 'r', encoding='utf-8')
    while True:
        line = file.readline()
        if not line:
            break
        print(line.strip())
    file.close()
except FileNotFoundError:
    print('bugsTOP100.txt 파일이 디스크에 없습니다.')
```

```python
try:
    with open('./output/bugsTOP100.txt', 'r', encoding='utf-8') as file:
        lines = file.readlines()
        for line in lines:
            print(line.strip())
except FileNotFoundError:
    print('bugsTOP100.txt 파일이 디스크에 없습니다.')
```
