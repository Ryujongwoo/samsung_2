# 서포트 벡터 머신(Support Vector Machine, SVM) 알고리즘

kNN 알고리즘은 주변의 가까운 이웃들에게 물어봐서 대다수의 답변과 동일하게 예측하는 방법이다. 만약에 3명에게 물어봤는데 2명이 거짓으로 대답한다면 예측은 틀리게 된다. kNN은 `지금 현재 이곳이 강남인가요? 강북인가요?`와 같이 질문한다면 SVM은 `지금 현재 이곳이 한강의 남쪽인가요? 북쪽인가요?`와 같이 질문해서 답변을 얻는 방식이다.

서포트 벡터라는 개념은 `결정 경계선`(한강)의 위치를 찾는데 사용되는 개념으로 한강은 현재 위치가 강남인지 강북인지 구분하는 결정 경계선이다. 서포트 벡터는 결정 경계선을 찾는 데 사용되는 2차원 공간상에 나타나는 데이터 포인트를 의미한다. 따라서, `서포트 벡터는 결정 경계선과 가장 가까이 맞닿은 데이터 포인트`를 의미한다.

서포트 벡터와 결정 경계선 사이의 거리를 `마진`이라하고 서포트 벡터 머신의 목표는 `마진을 최대로 하는 결정 경계를 찾는 것`이다.

서포트 벡터 머신에서 약간의 `오차를 허용`를 허용하기 위해서 `비용(C)`이라는 변수를 사용한다. 비용이 낮을(오차가 작을)수록 마진을 최대한 높이고 학습 에러율을 증가시시키는 방향으로 결정 경계선을 만들고 비용이 높을(오차가 큰) 마진은 최대한 낮추고 학습 에러율을 감소시키는 방향으로 결정 경계를 만든다. 비용이 너무 낮으면 과적합의 위험이 있으니 적절한 비용값을 찾는 과정이 상당히 중요하다.

## 서포트 벡터 머신 알고리즘에 사용할 데이터를 준비한다.

```python
raw_data = datasets.load_wine()

xData = raw_data.data
yData = raw_data.target

x_train, x_test, y_train, y_test = train_test_split(xData, yData, random_state=0)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

## 서포트 벡터 머신 모델을 만들고 학습 시킨다.

서포트 벡터 머신 알고리즘을 사용하기 위해 SVC를 import 한다.

```python
from sklearn.svm import SVC
```

서포트 벡터 머신의 모델을 만든다.

학습된 모델을 테스트 데이터로 예측할 때 predict() 메소드를 사용하면 아무런 문제가 없지만 predict_proba() 메소드를 사용하면 에러가 발생되므로 predict_proba() 메소드를 사용해야 한다면 `probability`를 `True`로 설정해야 한다.

```python
model = SVC(kernel='rbf', probability=True)
```

서포트 벡터 머신 모델을 학습시키기 위해서 fit() 메소드에 `학습 데이터(x_train)`와 `학습 데이터에 따른 레이블(y_train)`을 순서대로 넘긴다.

```python
model.fit(x_train, y_train)
```

## 학습된 서포트 벡터 머신 모델을 테스트 데이터로 예측한다.

predict() 메소드에 `테스트 데이터(x_test)`를 넘겨서 결과를 예측한다. `예측값`을 계산한다.

```python
predict = model.predict(x_test)
print(predict)
```

predict_proba() 메소드는 `테스트 데이터(x_test)`를 넘겨서 결과를 각 클래스에 `속할 확률`로 예측한다.

```python
predict_proba = model.predict_proba(x_test)
print(predict_proba)
```

## 모델을 평가한다.

```python
classification = classification_report(y_test, predict, target_names=raw_data.target_names)
print(classification)
```
