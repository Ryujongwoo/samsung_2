# 파이썬 설치

파이썬을 설치할 폴더(`python`)과 우리가 작업하는 소스 파일을 모아놓을 폴더(`workspace`)를 만든다.

<img src="./images/python_install_1.png" width="800" align="left" />

파이썬 다운로드를 검색한다.

<img src="./images/python_install_2.png" width="800" align="left" />

파이썬 프로그램을 다운받을 링크를 클릭한다.

<img src="./images/python_install_3.png" width="800" align="left" />

파이썬 3.11 버전을 설치하기 위해서 설치 프로그램이 있는 링크를 클릭한다.

<img src="./images/python_install_4.png" width="800" align="left" />

파이썬 설치 프로그램을 클릭해서 다운로드 받는다.

<img src="./images/python_install_5.png" width="800" align="left" />

다운로드 받은 설치 프로그램을 더블클릭한다.

<img src="./images/python_install_6.png" width="800" align="left" />

Add python.exe to PATH에 체크하고 Customize installation를 클릭한다.

<img src="./images/python_install_7.png" width="800" align="left" />

<img src="./images/python_install_8.png" width="800" align="left" />

Browse 버튼을 클릭해서 파이썬 프로그램을 설치할 폴더를 선택하고 Install 버튼을 클릭해서 파이썬 프로그램을 설치한다.

<img src="./images/python_install_9.png" width="800" align="left" />

<img src="./images/python_install_10.png" width="800" align="left" />

# D2Coding 글꼴 설치 방법

D2Coding 글꼴이 있는 github 사이트를 검색한다.

<img src="./images/python_install_22.png" width="1200" align="left" />

D2Coding 글꼴이 있는 fonts/ttf 폴더를 클릭한다.

<img src="./images/python_install_23.png" width="1200" align="left" />

D2Coding-Bold.ttf와 D2Coding-Regular.ttf를 다운로드받아서 파일 이름위에서 마우스 오른쪽 버튼을 누른 후 설치를 클릭한다.  
설치 메뉴가 보이지 않으면 파일을 복사해서 Windows 폴더 아래의 Fonts 폴더에 복사해 넣어도 설치된다.
