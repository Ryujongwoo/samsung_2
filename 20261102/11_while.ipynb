# 제어문(반복문) - while

while 조건식:  
&nbsp;&nbsp;&nbsp;&nbsp;조건식이 참인 동안 실행할 문장  
&nbsp;&nbsp;&nbsp;&nbsp;...

while은 조건이 참인 동안 반복하고 조건이 거짓이 되는 순간 반복을 종료한다.  
while은 최초 진입 조건이 거짓일 경우 while 블록을 한 번도 실행하지 않는다.  
while은 반복을 하기는 해야겠는데 몇 번 반복해야하는지 모를때 사용하고 for는 반복 횟수를 알고있을 때 사용한다.

## 1에서 10까지의 합계

```python
total = 0 # 1 ~ 10의 합계를 누적할 변수
for i in range(1, 11):
    total = total + i
print(total)
```

```python
i = 0
total = 0
while i < 10:
    i += 1
    total += i
# ===== while
print(total)
```

```python
i = 0
total = 0

# 의도적으로 무한 루프를 발생시키려면 조건을 항상 참이되도록 하면된다.
# 의도적으로 발생시킨 무한 루프는 반드시 break 명령을 이용해서 무한 루프를 탈출시키는 처리를 해야한다.
while True:
    i += 1
    total += i
    if i == 10:
        break
    # ===== if
# ===== while
print(total)
```

```python
while True:
    while True:
        print('=' * 40)
        print(' 1.입력  2.보기  3.수정  4.삭제  5.종료 ')
        print('=' * 40)
        menu = int(input('원하는 메뉴를 입력하고 enter 키를 누르세요: '))

        # menu에 저장된 값이 1~5 사이라면 무한 루프를 탈출한다.
        # if menu >= 1 and menu <= 5:
        # if 1 <= menu <= 5:
        if menu in [1, 2, 3, 4, 5]:
            break
        else:
            print('메뉴는 1~5 사이의 값만 입력해야 합니다.')
        # ===== if
    # ===== while True:

    # 여기까지 넘어왔다면 menu에는 1~5 사이의 정수가 입력된 것이다.
    if menu == 1:
        print('입력 작업 실행')
    elif menu == 2:
        print('보기 작업 실행')
    elif menu == 3:
        print('수정 작업 실행')
    elif menu == 4:
        print('삭제 작업 실행')
    else:
        break
    # ===== if
# ===== while
print('프로그램을 종료합니다. 바이바이~~~~~')
```
