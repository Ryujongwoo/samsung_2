## range() 함수

range() 함수는 초기치, 최종치, 증가치에 따른 숫자 리스트를 만든다.  
range() 함수는 초기치 부터 시작해서 증가치 만큼 증가하며 최종치를 넘어가지 않는 숫자 리스트를 만든다.

`range([초기치, ]최종치[, 증가치])` []로 묶어놓은 내용은 필요에 따라서 생략할 수 있다.  
초기치와 증가치를 생략하면 초기치의 기본값은 `0`이고 증가치의 기본값은 `1`이 사용된다.  
즉, range(0, 10, 1)와 range(10)은 같은 표현이다.

```python
print(range(0, 10, 1), range(10))
```

range() 함수가 실행된 결과를 리스트로 확인하려면 리스트 생성자 함수 list()를 사용한다.

```python
print(list(range(0, 10, 1)))
print(list(range(10)))
print(list(range(0, 10, 2)))
print(list(range(1, 10, 2)))
print(list(range(9, -1, -1)))
```

```python
for i in [1, 2, 3, 4, 5]:
    print(i)
for i in range(1, 6):
    print(i)
```

## 1에서 10까지의 합계

```python
total = 0 # 1 ~ 10의 합계를 누적할 변수
for i in range(1, 11):
    total = total + i
print(total)
```

## 주사위를 10번 굴려서 출현한 눈의 개수를 센다. => 변수 사용

```python
import random
```

```python
number_1 = 0; number_2 = 0; number_3 = 0; number_4 = 0; number_5 = 0; number_6 = 0

for i in range(1, 11):
    dice = random.randint(1, 6)
    # 주사위를 굴려서 발생한 눈의 개수를 증가시킨다.
    if dice == 1:
        # number_1 = number_1 + 1
        number_1 += 1
    elif dice == 2:
        number_2 += 1
    elif dice == 3:
        number_3 += 1
    elif dice == 4:
        number_4 += 1
    elif dice == 5:
        number_5 += 1
    else:
        number_6 += 1
    # ===== if
# ===== for

# 주사위 눈의 개수를 출력한다.
print(f'1의 개수: {number_1}')
print(f'2의 개수: {number_2}')
print(f'3의 개수: {number_3}')
print(f'4의 개수: {number_4}')
print(f'5의 개수: {number_5}')
print(f'6의 개수: {number_6}')
```

## 주사위를 10번 굴려서 출현한 눈의 개수를 센다. => 리스트 사용

```python
# 1. 주사위 눈의 개수를 기억할 크기가 6인 0으로 초기화된 리스트를 만든다.
# number = [0, 0, 0, 0, 0, 0]
# number = [0] * 6
# 2. 빈 리스트를 선언하고 append() 메소드로 0을 넣어준다.
# number = []
# for i in range(6):
    # number.append(0)
# 3. 컴프리핸션을 사용해서 리스트를 만든다.
number = [0 for i in range(6)]
# print(number)

for i in range(1, 11):
    # dice = random.randint(1, 6)
    # number[dice - 1] += 1
    # number[random.randint(1, 6) - 1] += 1
    number[random.randrange(1, 6)] += 1

for i in range(6):
    print(f'{i + 1}의 개수: {number[i]}')  
```

## 리스트와 함께 사용하는 for

```python
a = ['Life', 'is', 'too', 'Good']
print(a)
print('-' * 80)

for i in a:
    print(i)
print('-' * 80)

for i in range(len(a)):
    print(i, a[i])
print('-' * 80)

for i in a[1:3]:
    print(i)
```

## 딕셔너리와 함께 사용하는 for

```python
a = {'apple': 1000, 'banana': 2000, 'melon': 10000, 'mango': 5000}
print(a)
print(a.keys())
print(a.values())
print(a.items())
```

```python
# for에서 딕셔너리 이름만 사용하면 딕셔너리이름.keys()와 같은 의미로 사용된다.
for i in a:
    print(i)
print('-' * 80)

for i in a.keys():
    print(i)
print('-' * 80)

for i in a.values():
    print(i)
print('-' * 80)

for i in a.items():
    print(i)
```

key를 이용해서 value 얻어오기

```python
for key in a.keys():
    print(a[key]) # 딕셔너리이름[key]
print('-' * 80)

for key in a.keys():
    print(a[key]) # 딕셔너리이름.get(key, default)
```

items를 이용해서 key와 value 얻어오기

```python
for item in a.items():
    print(item)
print('-' * 80)

for item in a.items():
    print('{}는(는) {:,}원 입니다.'.format(item[0], item[1]))
print('-' * 80)

for key, value in a.items():
    print(f'{key}는(는) {value:,}원 입니다.')
```

## enumerate() 함수

enumerate() 함수는 인수로 지정된 리스트, 튜플, set, 딕셔너리의 index와 데이터를 쌍으로 묶은 튜플을 리턴한다.

```python
a = ['Life', 'is', 'too', 'Good']
print(a)
print(enumerate(a))
print(list(enumerate(a)))

for index, data in enumerate(a):
    print(index, data)
```

## zip() 함수

zip() 함수는 서로 다른 데이터가 저장된 객체를 묶어서 반복 가능한 형태로 만든다.

```python
a = ['java', 'jsp', 'python']
b = [100, 95, 98]

print(zip(a, b))
print(list(zip(a, b)))

for i in zip(a, b):
    print(i)
```
