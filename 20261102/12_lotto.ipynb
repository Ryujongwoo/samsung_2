# 로또 번호 생성기

로또 번호는 난수를 발생시켜 처리해야하므로 random 모듈을 import 한다.

```python
import random
```

## 리스트를 이용한 로또 번호 생성기

```python
# 로또 1등 번호를 기억할 빈 리스트를 만든다.
lotto = []

# 랜덤을 실행하면 중복되는 번호가 몇 번 나올지 모르기때문에 while 반복을 무한 루프를 돌려서 처리한다.
while True:
    # 1 ~ 45 사이의 난수를 발생시킨다.
    number = random.randint(1, 45)
    # print(number)
    # 1등 번호를 기억할 리스트에 중복되는 번호가 없다면 난수를 추가한다.
    if number not in lotto:
        lotto.append(number)
    # 리스트에 중복되지 않은 숫자가 7개가 저장되면 무한 루프를 탈출한다.
    if len(lotto) == 7:
        break
# ===== while
print('1등 번호: {}, 보너스 번호: {}'.format(lotto[0:6], lotto[6]))
print('1등 번호: {}, 보너스 번호: {}'.format(lotto[:-1], lotto[-1]))
print('1등 번호: {}, 보너스 번호: {}'.format(sorted(lotto[:-1]), lotto[-1]))
print('1등 번호: {}, 보너스 번호: {}'.format(sorted(lotto[:-1], reverse=True), lotto[-1]))
```

프로그램을 인수로 지정한 시간 만큼 멈추는 sleep() 메소드를 사용하기 위해서 time 모듈을 import 한다.

```python
import time
```

```python
print('1등 번호: ', end='')
for i in lotto[:-1]:
    print(i, end=', ')
    # time 모듈의 sleep() 메소드를 사용하면 인수로 지정한 시간만큼 프로그램을 멈출 수 있다. 시간은 초 단위로 지정한다.
    time.sleep(1)
print('보너스 번호: ', lotto[-1])
```

## set을 이용한 로또 번호 생성기

```python
# 로또 1등 번호를 기억할 빈 set을 만든다.
lotto = set()

# 1등 번호
while True:
    number = random.randint(1, 45)
    # 리스트는 중복을 확인하는 if가 필요했지만 set은 중복되는 데이터를 허용하기 않기 때문에 중복을 확인하는 if가 필요없다.
    lotto.add(number)
    # print(number)
    # 리스트에 중복되지 않은 숫자가 6개가 저장되면 무한 루프를 탈출한다.
    if len(lotto) == 7:
        break
    # ===== if
# ===== while
print(f'1등 번호: {lotto}')

# 보너스 번호
while True:
    bonus = random.randint(1, 45)
    if bonus not in lotto:
        break
    # ===== if
# ===== while
print('1등 번호: {}, 보너스 번호: {}'.format(lotto, bonus))
```

## 로또 추첨 방송처럼 로또 번호 생성기를 구성한다.

%%time 매직 명령은 jupyter nbclassic의 한 셀이 실행되는 시간을 출력한다.

```python
%%time

# 로또 추첨기로 사용할 리스트를 만들고 1 ~ 45로 초기화시킨다.
lotto = [i for i in range(1, 46)]
# print(lotto)

# 섞기전 상태를 한 줄에 숫자 10개씩 출력한다.
for i in range(len(lotto)):
    # 02d: 정수 2자리로 출력한다. 남는 자리는 0을 채워서 출력한다.
    print('{:02d}'.format(lotto[i]), end=' ')
    if (i + 1) % 10 == 0:
        print() # 아무것도 출력하지 않고 줄을 바꾼다.
    # ===== if
# ===== for
print('\n' + '=' * 29 + ' 섞기전')

# 섞는다.
'''
# lotto[0]와(고정) lotto[1] ~ lotto[44] 중에서 랜덤한 위치의 값을 선택해서 두 기억 장소의 값을 교환한다.
for i in range(100000):
    # 1 ~ 44 사이에서 랜덤한 값을 발생시킨다.
    r = random.randint(1, 44)
    # lotto[0]와 lotto[r]에 저장된 값을 교환한다. => 모든 프로그래밍 언어에서 공통적으로 사용하는 방법
    # temp = lotto[0]
    # lotto[0] = lotto[r]
    # lotto[r] = temp
    lotto[0], lotto[r] = lotto[r], lotto[0]
'''
# shuffle() 메소드는 인수로 지정된 리스트의 내용을 랜덤하게 섞는다. => 반복 처리보다 속도가 월등하게 빠르다.
random.shuffle(lotto)

# 섞은후 상태를 한 줄에 숫자 10개씩 출력한다.
for i in range(len(lotto)):
    print('{:02d}'.format(lotto[i]), end=' ')
    if (i + 1) % 10 == 0:
        print()
print('\n' + '=' * 29 + ' 섞은후')

# 섞은 결과에서 앞의 6개가 1등 번호이고 6번째 인덱스의 값이 보너스 번호이다.
print('1등 번호: {}, 보너스 번호: {}'.format(lotto[0:6], lotto[6]))
```

```python
# sample() 메소드는 첫 번째 인수로 지정된 리스트에서 두 번째 인수로 지정된 숫자 만큼의 데이터를 무작위로 꺼내온다.
print(random.sample(lotto, 7))
```

## 미국 복권(power ball) 번호 생성기

1 ~ 69 사이에서 5개, 1 ~ 26 사이에서 1개

```python
lotto = [i for i in range(1, 70)]
random.shuffle(lotto)
print('1등 번호: {}, 파워볼: {}'.format(lotto[0:5], random.randint(1, 26)))
```

```python
print('1등 번호: {}, 파워볼: {}'.format(random.sample(lotto, 5), random.randint(1, 26)))
```

## 로또 복권 구매 금액을 키보드로 입력받아 입력받은 금액 만큼 로또 번호를 생성한다.

```python
lotto = [i for i in range(1, 46)]

# 로또 복권 구매 금액을 입력받는다.
money = int(input('로또 복권 구매 금액을 입력하세요: '))

# 로또 구매 금액만큼 반복하며 로또 번호를 생성한다.
for i in range(money // 1000):
    print('{:3d} game: '.format(i + 1), end='')
    random.shuffle(lotto)
    result = lotto[:6]
    result.sort()
    for j in range(6):
        print('{:02d} '.format(result[j]), end='')
    # ===== for j
    print()
# ===== for i
```

```python
money = int(input('로또 복권 구매 금액을 입력하세요: '))
for i in range(money // 1000):
    print('{:3d} game: '.format(i + 1), end='')
    result = random.sample(lotto, 6)
    result.sort()
    for j in range(6):
        print('{:02d} '.format(result[j]), end='')
    print()
```
