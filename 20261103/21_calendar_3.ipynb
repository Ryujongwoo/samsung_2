# 소스 파일과 module이 다른 폴더에 있는 경우 import 하기

import 할 때 `import calendarModule` 또는 `import calendarModule as cm`와 같이 module 이름만 사용하면 현재 작업중인 소스 프로그램의 위치한 경로에 있는 module을 import 시켜서 사용한다는 의미이다.  
import 할 module이 현재 작업중인 소스 프로그램이 위치한 폴더에 존재하지 않고 다른 폴더에 위치한 경우 import 하는 방법을 알아본다.

## module이 소스 프로그램이 위치한 폴더의 하위 폴더에 있는 경우

import 할 모듈이 소스 프로그램이 위치한 폴더의 하위 폴더에 있는 경우 import 할 모듈이 위치한 하위 폴더 이름 뒤에 `.`을 찍고 import 하면 된다.

```python
import lib.calModule as cm
```

```python
year, month = map(int, input('달력을 출력할 년, 월을 입력하세요: ').split())

print('=' * 28)
print('         {:4d}년{:2d}월'.format(year, month))
print('=' * 28)
print(' 일  월  화  수  목  금  토 ')
print('=' * 28)

for i in range(cm.weekDay(year, month, 1)):
    print('    ', end='')

for i in range(1, cm.lastDay(year, month) + 1):
    print(' {:2d} '.format(i), end='')
    if cm.weekDay(year, month, i) == 6 and i != cm.lastDay(year, month):
        print()
print('\n' + '=' * 28)
```

## module에서 특정 기능만 import 해서 사용하기

module의 특정 기능만 import 해서 사용하려면 아래와 같이 import 시켜서 사용한다.  
`from 모듈이름 import 함수이름` => from 다음의 module에서 import 다음의 함수만 import 시켜서 사용할 수 있다.

```python
# from calendarModule import weekDay # calendarModule module의 weekDay 함수만 import 된다.
# from calendarModule import lastDay # calendarModule module의 lastDay 함수만 import 된다.

# weekDay 함수와 lastDay 함수는 모두 calendarModule의 함수이므로 아래와 같이 사용할 수 있다.
from calendarModule import weekDay, lastDay
```

from을 이용해서 특정 함수만 import 시키면 module 이름을 사용하지 않고 함수 이름만 사용할 수 있다.

```python
year, month = map(int, input('달력을 출력할 년, 월을 입력하세요: ').split())

print('=' * 28)
print('         {:4d}년{:2d}월'.format(year, month))
print('=' * 28)
print(' 일  월  화  수  목  금  토 ')
print('=' * 28)

for i in range(weekDay(year, month, 1)):
    print('    ', end='')

for i in range(1, lastDay(year, month) + 1):
    print(' {:2d} '.format(i), end='')
    if weekDay(year, month, i) == 6 and i != lastDay(year, month):
        print()
print('\n' + '=' * 28)
```
