인수로 년도를 넘겨받아 윤년, 평년을 판단해서 윤년이면 True, 평년이면 False를 리턴하는 isLeapYear(year) 함수를 만든다.

```python
def isLeapYear(year):
    return year % 4 == 0 and year % 100 != 0 or year % 400 == 0
```

인수로 년, 월을 입력받아 그 달의 마지막 날짜를 리턴하는 lastDay(year, month) 함수를 만든다.

```python
def lastDay(year, month):
    # 각 달의 마지막 날짜를 기억하는 리스트를 만든다.
    m = [31, 0, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    # 인수로 넘겨받은 년도에 해당되는 2월의 마지막 날짜를 확정한다.
    '''
    if isLeapYear(year):
        m[1] = 29
    else:
        m[1] = 28
    '''
    # m[1]에 29을 넣어줄거야 만약에 윤년이 아니면 28을 넣어줄거야 
    m[1] = 29 if isLeapYear(year) else 28
    # 마지막 날짜를 리턴한다.
    return m[month - 1]
```

인수로 년, 월, 일을 입력받아 1년 1월 1일부터 지나온 날짜의 합계를 계산해서 리턴하는 totalDay(year, month, day) 함수를 만든다.

```python
def totalDay(year, month, day):
    # 인수로 넘어온 년도가 다 지나가지 않았으므로 1년 1월 1일부터 전년도 12월 31일까지 지난 날짜를 계산한다.
    days = (year - 1) * 365 + (year - 1) // 4 - (year - 1) // 100 + (year - 1) // 400
    # 전년도 12월 31일까지 지난 날짜에 전달까지 지난 날짜를 더한다.
    for i in range(1, month):
        days += lastDay(year, i)
    # 전달까지 지난 날짜에 일을 더해서 리턴시킨다.
    return days + day
```

인수로 년, 월, 일을 입력받아 요일을 계산해서 숫자로 리턴하는 weekDay(year, month, day) 함수를 만든다.  
weekDay() 함수의 실행 결과가 1이면 월요일, 2면 화요일, 3은 수요일, 4는 목요일, 5는 금요일, 6은 토요일, 0은 일요일을 의미한다.

```python
def weekDay(year, month, day):
    # 1년 1월 1일부터 요일을 계산할 날짜까지 지난 날짜를 7로 나눈 나머지를 리턴시킨다.
    return totalDay(year, month, day) % 7
```
