모든 프로그래밍 언어는 함수나 메소드의 실행 결과를 return 뒤에 딱 1개만 붙여서 return 시킬 수 있다. => 파이썬도 예외는 아니다.

```python
def swap(a, b):
    temp = a
    a = b
    b = temp
    # 데이터를 ','로 구분해서 나열하면 ()로 묶지않아도 튜플로 인식한다.
    # return을 ()를 사용하지 않고 튜플 형태로 return 시키면 return 값이 여러개인 것 같은 효과를 낸다.
    return a, b # (a, b) 형태의 튜플을 리턴시킨다.
```

```python
a = 3
b = 4
print(f'a = {a}, b = {b}')

[a, b] = [b, a]
print(f'a = {a}, b = {b}')
(a, b) = (b, a)
print(f'a = {a}, b = {b}')
a, b = b, a
print(f'a = {a}, b = {b}')

result = swap(a, b)
print(result)
print(type(result))
print(f'a = {result[0]}, b = {result[1]}')

a, b = swap(a, b)
print(f'a = {a}, b = {b}')
```

함수 내부의 변수와 함수 외부의 변수는 이름이 같더라도 서로 독립적인 기억 공간을 가진다.

```python
def func9(var1):
    var1 += 1
    print(f'func9() 함수 내부의 변수 var1: {var1}')
```

```python
var1 = 1
func9(var1)
print(f'func9() 함수 외부의 변수 var1: {var1}')
```

## global 예약어를 사용하면 함수 내부에서 함수 외부의 변수를 사용할 수 있다.

C/C++ 언어의 call by reference(참조(주소)에 의한 호출) 방식처럼 사용된다. => side effect(부작용) 현상이 발생될 수 있으니 주의해서 사용해야 한다.

```python
def func10():
    # global로 선언한 변수는 함수의 인수로 사용하면 안된다.
    # global로 선언한 변수는 함수 내부에서 값을 변경하면 함수 외부에서도 값이 변경된다.
    global var2 # 함수 외부에서 선언된 변수를 함수 내부에서 사용한다고 선언한다.
    var2 += 1
    print(f'func10() 함수 내부의 변수 var2: {var2}')
```

```python
var2 = 1
func10()
print(f'func10() 함수 외부의 변수 var2: {var2}')
```
