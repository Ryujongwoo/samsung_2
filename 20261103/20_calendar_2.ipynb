# 사용자 정의 모듈 import

사용자가 만든 파이썬 파일을 import 시켜서 사용할 수 있다.  
import 할 모듈의 이름은 반드시 문자로 시작해야 하고 현재 작성중인 소스 파일 이름과 달라야 한다.

calendarModule.py 파일의 모든 함수가 import 된다.

```python
import calendarModule
```

```python
year, month = map(int, input('달력을 출력할 년, 월을 입력하세요: ').split())

print('=' * 28)
print('         {:4d}년{:2d}월'.format(year, month))
print('=' * 28)
print(' 일  월  화  수  목  금  토 ')
print('=' * 28)

for i in range(calendarModule.weekDay(year, month, 1)):
    print('    ', end='')

for i in range(1, calendarModule.lastDay(year, month) + 1):
    print(' {:2d} '.format(i), end='')
    if calendarModule.weekDay(year, month, i) == 6 and i != calendarModule.lastDay(year, month):
        print()
print('\n' + '=' * 28)
```

`import calendarModule` 형태로 import 하면 라이브러리 이름이 너무 길어서 사용하기 불편하기 때문에 `as`를 사용해서 module의 별명을 지정할 수 있다.  
as로 별명을 지정하면 `calendarModule`는 사용할 수 없고 `cm`만 사용해야 한다.

```python
import calendarModule as cm
```

```python
year, month = map(int, input('달력을 출력할 년, 월을 입력하세요: ').split())

print('=' * 28)
print('         {:4d}년{:2d}월'.format(year, month))
print('=' * 28)
print(' 일  월  화  수  목  금  토 ')
print('=' * 28)

for i in range(cm.weekDay(year, month, 1)):
    print('    ', end='')

for i in range(1, cm.lastDay(year, month) + 1):
    print(' {:2d} '.format(i), end='')
    if cm.weekDay(year, month, i) == 6 and i != cm.lastDay(year, month):
        print()
print('\n' + '=' * 28)
```
