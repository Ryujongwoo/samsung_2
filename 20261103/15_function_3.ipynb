## 함수로 데이터를 전달할 때 데이터가 전달될 인수를 지정할 수 있다.

```python
def func3(a, b, c):
    print('{} + {} + {} = {}'.format(a, b, c, a + b + c))
```

데이터를 함수로 전달할 때 인수를 지정하지 않으면 처음부터 차례대로 데이터가 함수로 전달된다.

```python
func3(100, 200, 300)
```

함수를 호출할 때 데이터가 전달될 인수를 지정할 수 있다.

```python
func3(a=100, b=200, c=300)
```

함수를 호출할 때 데이터가 전달될 인수를 지정하면 가인수의 순서와 관계없이 사용할 수 있다.

```python
func3(b=100, c=200, a=300)
```

```python
# 인수를 지정해서 함수를 호출할 때 가인수 목록에 없는 인수를 지정하면 에러가 발생된다.
func3(b=100, c=200, d=300)
```

## 디폴트(기본값) 인수를 사용하는 함수

실인수과 가인수의 이름은 반드시 같을 필요는 없지만 인수의 개수는 반드시 같아야 한다. => 디폴트 인수를 사용하면 달라도 상관없다.  
함수가 호출될 때 데이터가 넘어오면 넘어온 데이터로 함수를 실행하고 데이터가 넘어오지 않으면 디폴트로 지정함 데이터로 함수를 실행한다.

```python
def func4(a = 8, b = 6):
# 일반 인수와 디폴트 인수를 함께 사용하려는 경우 일반 인수를 모두 선언한 후 디폴트 인수를 선언해야 한다.
# def func4(a = 8, b): # 에러
# def func4(a, b = 6): # 정상
    print('{} + {} = {}'.format(a, b, a + b))
```

```python
func4(1, 2)
```

```python
func4()
```

```python
func4(9)
```

```python
func4(b=9)
```

데이터가 넘어오면 1부터 넘어온 데이터까지 합계를 계산하고 데이터가 넘어오지 않으면 1에서 100까지 합계를 계산하는 함수

```python
def func5(n = 100):
    '''
    total = 0
    for i in range(1, n + 1):
        total += i
    return total
    '''
    return (n + 1) * (n // 2)
```

```python
print(func5(10))
```

```python
print(func5())
```
