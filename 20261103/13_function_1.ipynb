# 함수

프로그램에서 중복해서 처리해야 할 부분을 별도로 작성하는 것을 말한다.

## 함수 실행 순서

프로그램을 실행하다 함수가 호출되면 호출된 함수로 제어가 넘어간다. => 함수가 실행된 후 복귀할 주소는 스택에 저장된다.  
호출된 함수가 정의된 기능을 모두 실행하거나 return 명령을 만나면 함수가 종료된다. => 스택에 저장된 복귀 주소를 참조한다. => 마지막에 호출된 함수가 먼저 종료된다.  
return이 있으면 return 뒤의 값을 가지고 함수가 호출된곳으로 돌아가고 return이 없거나 return만 있으면 None을 가지고 돌아간다.

## 함수의 형식

def 함수이름([인수, ...]): # 인수는 필요없으면 생략할 수 있다.  
&nbsp;&nbsp;&nbsp;&nbsp;함수가 실행할 문장  
&nbsp;&nbsp;&nbsp;&nbsp;...  
&nbsp;&nbsp;&nbsp;&nbsp;[return 함수의 실행 결과 값] # retun도 생략할 수 있다. 생략하면 None이 리턴된다.  

```python
# 가장 일반적으로 사용하는 함수의 모양
# a, b 2개의 데이터를 인수로 넘겨받아 덧셈한 결과를 리턴하는 함수
def add(a, b): # 호출당하는 함수의 인수를 가인수라 부른다.
    print('add(a, b) 함수가 호출됨')
    result = a + b
    return result
```

실인수과 가인수의 이름은 반드시 같을 필요는 없지만 인수의 개수는 반드시 같아야 한다.  
실인수와 가인수는 서로 독립적인 메모리 공간을 사용하기 때문에 가인수의 값이 변경되더라도 실인수의 값은 변경되지 않는다.

```python
print(f'함수 실행 결과: {add(1, 2)}') # 실행할 함수를 호출한다. 함수를 호출하는 부분의 인수를 실인수라 부른다.
```
