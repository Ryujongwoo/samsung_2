# 튜플

튜플은 리스트와 거의 비슷하지만 약간의 차이가 있다.  
리스트는 데이터를 대괄호(`[]`)를 사용해서 묶어주고 튜플은 데이터를 소괄호(`()`)를 사용해서 묶어준다.  
리스트는 데이터의 수정 및 삭제가 가능하지만 튜플은 데이터의 `수정 및 삭제가 불가능`하다.

```python
# 빈 튜플 만들기
a = ()
print(a)
print(type(a))

b = tuple() # 튜플 생성자 함수
print(b)
print(type(b))
```

```python
# 데이터가 1개인 리스트
a = [1]
print(a)

# 데이터가 1개인 튜플
b = (1)
print(type(b))
print(b)

# 데이터가 1개인 튜플을 만들려면 ()안의 데이터 뒤에 반드시 ','를 입력해야 한다.
# 데이터가 1개인 튜플을 만들기 위해서 ()만 사용하면 파이썬은 ()를 튜플로 취급하지 않고 연산식의 우선순위를 의미하는 ()로 취급한다.
c = (1,)
print(type(c))
print(c)
```

```python
# 초기치를 가지는 튜플 만들기
a = (1, 2, 3)
print(a)

b = ('a', 'b', 'c')
print(b)

c = (100, 'aaa', 200, 'bbb')
print(c)

d = ((100, 'aaa'), (200, 'bbb'))
print(d)

e = ([100, 'aaa'], [200, 'bbb'])
print(e)
```

## 튜플은 리스트와 똑같이 인덱싱과 슬라이싱이 가능하다.

```python
a = (1, 2, 3, 4, 5)
print(a)

# 인덱싱
print(a[2])
print(a[-1])

# 슬라이싱
print(a[1:4])
print(a[0:3])
print(a[:3])
print(a[2:5])
print(a[2:])
```

## 튜플에 사용하는 연산자

```python
a = (1, 2, 3)
b = (1, 2, 3, 4)

print(a + b)
print(a * 3)
```

## 튜플은 리스트와 달리 수정과 삭제가 불가능하다.

```python
a = (1, 2, 3, 4, 5)
a[0] = 999
```

```python
del a[0]
```

## 데이터를 ','로 구분해서 나열하면 튜플로 인식된다.

```python
a = 1, 2, 3, 4, 5
print(a)
```

# 두 기억장소에 저장된 데이터 교환하기

```python
a = 3
b = 4
print(f'a = {a}, b = {b}') # a = 3, b = 4

# 모든 언어에서 공통적으로 사용할 수 있는 방법
temp = a
a = b
b = temp
print(f'a = {a}, b = {b}') # a = 4, b = 3

# 파이썬에서만 사용할 수 있는 방법 - 리스트 사용
[a, b] = [b, a]
print(f'a = {a}, b = {b}') # a = 3, b = 4

# 파이썬에서만 사용할 수 있는 방법 - 튜플 사용
(a, b) = (b, a)
print(f'a = {a}, b = {b}') # a = 4, b = 3

a, b = b, a
print(f'a = {a}, b = {b}') # a = 3, b = 4
```
