# 리스트

리스트는 하나의 이름으로 여러개의 기억장소를 만들어 사용한다.

```python
# 빈 리스트 만들기
a = []
print(a)
print(type(a))

b = list() # 리스트 생성자 함수
print(b)
print(type(b))
```

```python
# 초기치를 가지는 리스트 만들기
a = [1, 2, 3]
print(a)

b = ['a', 'b', 'c']
print(b)

c = [100, 'aaa', 200, 'bbb']
print(c)

d = [[100, 'aaa'], [200, 'bbb']]
print(d)
```

## 리스트의 인덱싱과 슬라이싱

```python
a = [1, 2, 3, 4, 5]
print(a)

# 인덱싱
print(a[2])
print(a[-1])

# 슬라이싱
print(a[1:4])
print(a[0:3])
print(a[:3])
print(a[2:5])
print(a[2:])
```

## 리스트에 사용하는 연산자

```python
a = [1, 2, 3]
b = [1, 2, 3, 4]

print(a + b) # '+' 연산자는 리스트를 연결한다.
print(a * 3) # '*' 연산자는 리스트를 지정한 횟수만큼 반복한다.
```

## 리스트에 저장된 데이터의 수정 및 삭제

```python
a = [1, 2, 3]
print(a)

# 리스트를 수정하려면 수정할 데이터가 저장된 오소를 인덱싱한 후 데이터를 넣어주면 된다.
a[1] = 5
print(a)

# del 명령으로 리스트의 특정 인덱스의 데이터를 삭제할 수 있다.
del a[1]
print(a)

# clear() 메소드로 리스트에 저장된 모든 데이터를 삭제할 수 있다.
a.clear()
print(a)

# del 명령으로 리스트 자체를 삭제할 수 있다.
del a
print(a)
```

# 리스트 메소드

```python
a = [1, 2, 3, 4, 5]

# len() 함수로 리스트에 저장된 데이터의 개수를 얻어온다.
print(len(a))

# append(데이터) 메소드로 리스트의 맨 뒤에 데이터를 추가할 수 있다.
print(a)
a.append(6)
print(a)

# insert(인덱스, 데이터) 메소드로 리스트의 특정 위치에 데이터를 삽입할 수 있다.
a.insert(2, 100)
print(a)
```

sort() 메소드로 리스트에 저장된 데이터를 정렬할 수 있다. 별도의 설정을 하지 않으면 오름차순(문자열은 가나다순)으로 정렬한다.

```python
a = [1, 5, 2, 3, 6, 4]
print(a)
a.sort()
print(a)

b = ['d', 'b', 'a', 'c']
print(b)
b.sort()
print(b)

c = ['홍길동', '임꺽정', '장길산', '일지매']
print(c)
c.sort()
print(c)

d = ['1', '010', '999', '50']
print(d)
d.sort()
print(d)
```

sort() 메소드에 `reverse=True`를 지정하면 내림차순으로 정렬된다.

```python
a = [1, 5, 2, 3, 6, 4]
print(a)
a.sort(reverse=True)
print(a)

b = ['d', 'b', 'a', 'c']
print(b)
b.sort(reverse=True)
print(b)

c = ['홍길동', '임꺽정', '장길산', '일지매']
print(c)
c.sort(reverse=True)
print(c)

d = ['1', '010', '999', '50']
print(d)
d.sort(reverse=True)
print(d)
```

리스트에 저장된 데이터의 타입이 2종류 이상이면 sort() 메소드를 실행하면 에러가 발생된다.

```python
a = [1, 'a', 2, 'b']
print(a)
a.sort()
print(a)
```

sort() 메소드를 실행하는 리스트는 리스트 원본 자체가 정렬되고 sorted() 함수는 함수를 실행하는 리스트는 건드리지 않고 정렬된 결과를 리턴한다.

```python
a = [1, 5, 2, 3, 6, 4]
print(a)
print(sorted(a))
print(a)
b = sorted(a)
print(b)
print(sorted(a, reverse=True))
```

reverse() 메소드는 리스트에 저장된 내용을 뒤집는다.

```python
a = [1, 5, 2, 3, 6, 4]
a.reverse()
print(a)

# 오름차순으로 정렬 후 리스트에 저장된 내용을 뒤집어서 내림차순으로 정렬한 결과를 만든다.
a.sort()
a.reverse()
print(a)
```

```python
a = [1, 5, 2, 3, 6, 4]
print(a)
print(a[::-1])
```

count() 메소드는 리스트에서 인수로 지정된 데이터의 출현 횟수를 얻어온다.

```python
a = [1, 2, 3, 1, 3, 1, 4, 1, 5]
print(a.count(1))
print(a.count(6))
```

index() 메소드는 리스트에서 인수로 지정된 데이터가 최초로 출현하는 인덱스를 얻어온다.

```python
a = [1, 2, 3, 1, 3, 1, 4, 1, 5]
print(a.index(3))
print(a.index(6))
```

remove() 메소드로 리스트에서 인수로 지정한 첫 번째 데이터를 제거한다.  
del 명령은 지정된 인덱스의 데이터를 제거하고 remove() 메소드는 지정된 데이터를 제거하는 차이점이 있다.

```python
a = [1, 2, 3, 1, 3, 1, 4, 1, 5]
print(a)
a.remove(3)
print(a)
a.remove(3)
print(a)
a.remove(3)
```

pop() 메소드는 별도의 설정이 없으면 리스트의 맨 마지막 데이터를 리턴 후 제거하고 인수로 인덱스를 지정하면 지정한 인덱스에 해당되는 데이를 리턴 후 제거한다.

```python
a = [1, 2, 3, 1, 3, 1, 4, 1, 5]
print(a)
print(a.pop())
print(a)
print(a.pop(6))
print(a)
```
