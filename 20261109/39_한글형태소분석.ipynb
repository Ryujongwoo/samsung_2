# 형태소 분석(morphological analysis)

형태소 분석은 자연 언어의 문장을 형태소라는 최소 단위로 분할하고 품사를 판별하는 작업이다.  
영문 형태소 분석은 형태소마다 띄어쓰기를 해서 문장을 구성하는 것이 기본이기 때문에 크게 어렵지 않으나 아시아 계열 언어 분석에는 많은 노력이 필요하다.  
대표적으로 문법 규칙에 의한 확률적 언어 모델을 사용하는 방법이 있는데 최근 확률적 언어 모델을 사용한 형태소 분석 많아서 정밀도가 높아졌다.  
어쨋거나 두 가지 품사 사전과 문법 사전을 기반으로 대조하면서 형태소 분석을 한다.

konlpy에 포함된 대부분의 분석기는 `java` 기반으로 동작하기 때문에, 라이브러리를 정상적으로 사용하려면 컴퓨터에 `JDK(Java Development Kit)가 설치`되어 있어야 하며 `환경 변수(JAVA_HOME)`가 올바르게 설정되어 있어야 한다.

한국어 형태소 분석 라이브러리(konlpy)를 사용하기 위해 라이브러리를 설치하고 import 한다.  
`!pip install konlpy`

konlpy 라이브러리가 테스트용으로 제공하는 말뭉치(corpus)를 불러온다.

```python
from konlpy.corpus import kolaw, kobill
```

fileids() 메소드를 실행하면 각 말뭉치가 저장된 파일의 이름을 얻어온다.

```python
print(kolaw.fileids())
print(kobill.fileids())
```

말뭉치가 저장된 텍스트 파일을 읽어오려면 open() 메소드로 파일을 열과 read() 메소드로 읽는다.

```python
kolaw_corpus = kolaw.open('constitution.txt').read()
print(kolaw_corpus[:40])
```

```python
kobill_corpus = kobill.open('1809890.txt').read()
print(kobill_corpus[:40])
```

## konlpy가 제공하는 형태소 분석기

Hannanum(한나눔): KAIST에서 개발한 오랜 전통의 한국어 분석기이다.  
Kkma(꼬마): 서울대학교에서 개발한 분석기로, 품사 태그가 매우 세밀하고 문장 분석 능력이 뛰어나지만 분석 속도가 다소 느린 편이다.  
Komoran(코모란): 대용량 텍스트 처리에 적합하도록 설계되어 속도가 안정적이고 오탈자에 강하다.  
Okt(Open Korea Text): 트위터에서 개발한 분석기로, 비교적 속도가 빠르고 인터넷 신조어, 오타, 일상어 처리에 강점이 있어 가장 대중적으로 많이쓰인다.  
Mecab(메캅): C++로 작성되어 속도가 매우 빠르고 가벼워 대용량 데이터 처리나 실시간 서비스에 자주 쓰입니다(단, 윈도우 환경에서는 설치가 다소 까다롭다).
