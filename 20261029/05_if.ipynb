## 관계 연산자
`<` : 작다, 미만  
`<=`: 작거나 같다, 이하  
`>` : 크다, 초과  
`>=`: 크거나 같다, 이상  
`==`: 같다  
`!=`: 같지 않다

## 논리 연산자

`and`: 논리곱, 두 조건이 모두 참일 경우에만 참, ~이고, ~이면서, ~중에서  
`or` : 논리합, 두 조건 중에서 한 개 이상 참일 경우에 참, ~또는, ~이거나   
`not`: 논리부정, 참은 거짓으로 거짓은 참으로 변경한다.

## 연산자 우선순위

단항 연산자 => 이항 연산자 => 삼항 연산자 => 대입 연산자(`=`와 같이 사용한다.)  
이항 연산자: 산술 연산자(** => *, /, //, % => +, -) => 관계 연산자 => 논리 연산자(not => and => or)

둘리 주민등록번호(830422-1185600)을 입력받아 성별을 판단한다.

```python
jumin = input('주민등록번호를 입력하세요: ')
print(jumin[7])

# 문자열로 비교
if jumin[7] == '1' or jumin[7] == '3' or jumin[7] == '5':
    print('남자')
else:
    print('여자')
    
if jumin[7] in ['1', '3', '5']:
    print('남자')
else:
    print('여자')
print('-' * 100)

# 숫자로 비교
if int(jumin[7]) == 1 or int(jumin[7]) == 3 or int(jumin[7]) == 5:
    print('남자')
else:
    print('여자')
    
if int(jumin[7]) in [1, 3, 5]:
    print('남자')
else:
    print('여자')
print('-' * 100)

# 성별을 2로 나눈 나머지를 계산해서 홀수면 남자, 짝수면 여자
if int(jumin[7]) % 2 == 1:
    print('남자')
else:
    print('여자')
```

키보드로 년도를 입력받아 윤년이가 평년인가 판단한다.  
년도가 4로 나눠 떨어지고(and) 100으로 나눠 떨어지지 않거나(or) 400으로 나눠 떨어지면 윤년, 그렇치 않으면 평년

```python
year = int(input('윤년인가 평년인가 판별할 년도를 입력하세요: '))

if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
    print('{}년은 윤년입니다.'.format(year))
else:
    print(f'{year}년은 평년입니다.')
```

가위, 바위, 보 게임 만들기

난수를 발생시키는 random 모듈을 import 한다.

```python
import random
```

```python
# random() 메소드는 0이상이고 1미만인 난수를 실수로 발생시킨다.
print(random.random())
# randrange(a, b) 메소는 a부터 b-1 사이의 난수를 정수로 발생시킨다.
print(random.randrange(1, 6))
# randint(a, b) 메소는 a부터 b 사이의 난수를 정수로 발생시킨다.
print(random.randint(1, 6))
```

```python
# 가위(1), 바위(2), 보(3)
cpu = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3): '))
print(f'cpu: {cpu}, user: {user}')

if (cpu == 1 and user == 2) or (cpu == 2 and user == 3) or (cpu == 3 and user == 1):
    print('이겼다.')
elif (cpu == 3 and user == 2) or (cpu == 1 and user == 3) or (cpu == 2 and user == 1):
    print('졌다.')
else:
    print('비겼다.')
```
