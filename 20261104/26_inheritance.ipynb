# class 상속

부모 클래스에서 정의한 모든 내용(멤버 변수, 함수)을 자식 클래스에서 별도로 정의하지 않아도 정의한 것 처럼 사용할 수 있는 것을 말한다.

## 부모(상위, 슈퍼, 기반) 클래스

```python
class Parent:
    def __init__(self, name='무명씨', gender=False):
        print('부모 클래스의 생성자 실행')
        self.name = name
        self.gender = gender
        
    def __str__(self):
        return '{}({})'.format(self.name, '남' if self.gender else '여')
```

부모로 사용할 클래스가 잘 생성되었나 확인한다.

```python
parent1 = Parent()
print(parent1)
parent2 = Parent('이몽룡', True)
print(parent2)
parent3 = Parent('성춘향', False)
print(parent3)
```

## 자식(하위, 서브, 파생) 클래스

상속의 형식

`class 자식클래스이름(부모클래스이름)`

부모 클래스(Parent)를 상속받아 자식 클래스(Child)를 만든다.

Parent 클래스로부터 __init__(self, name='무명씨', gender=False), __str__(self), self.name, self.gender를 상속받았으므로 별도의 선언없이도 사용할 수 있다.  

```python
class Child(Parent):
    # 함수 재정의, Override
    # 부모 클래스에서 상속받은 함수의 기능만으로는 자식 클래스에서 사용하기에 불충분하기 때문에 자식 클래스에서 다시 정의하는 것을 말한다.
    def __init__(self, name='무명씨', gender=False, age=0, nickname='없음'):
        # 부모 클래스에서 상속받은 멤버 변수는 부모 클래스의 생성자를 호출해서 초기화시킨다.
        # super()는 부모 클래스를 의미한다.
        super().__init__(name, gender)
        print('자식 클래스의 생성자 실행')
        # 자식 클래스에서 별도로 정의한 멤버 변수들을 초기화시킨다.
        self.age = age
        self.nickname = nickname

    def __str__(self):
        return '{}({}) - {}, {}'.format(self.name, '남' if self.gender else '여', self.age, self.nickname)
```

```python
child = Child('홍길동', True, 20, '도적넘')
print(child)
print(child.name)
print(child.gender)
print(child.age)
print(child.nickname)
```
