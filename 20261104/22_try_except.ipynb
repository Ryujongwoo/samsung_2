# 에러 처리

프로그램을 작성할 잘못동작되는 것을 방지하기 위해서 에러가 발생된다. => 에러가 발생되면 프로그램이 멈춘다.  
try ~ except를 사용하면 에러가 발생되었을때 별도의 처리를 하거나 에러를 무시하고 프로그램을 실행할 수 있다.  

에러가 발생될 것으로 예상되는 문장들은 try 블록에 코딩하고 try 블록을 실행하다가 에러가 발생하면 try 블록에 남아있는 문장을 실행하지 않고 except 블록에 코딩된 문장을 실행한다.  
에러 발생 여부와 관계없이 무조건 실행해야 할 문장이 있다면 finally 블록에 코딩한다. => finally 블록은 선택사항 이다.

===============================================================================================================================================

try: # 필수  
&nbsp;&nbsp;&nbsp;&nbsp;에러가 발생될것으로 예상되는 문장  
&nbsp;&nbsp;&nbsp;&nbsp;...  
except [에러 이름 [as 변수 이름]]: # 필수, 에러 이름과 변수 이름은 생략할 수 있다.  
&nbsp;&nbsp;&nbsp;&nbsp;에러가 발생되면 실행항 문장  
&nbsp;&nbsp;&nbsp;&nbsp;...  
finally: # 선택, 필요한 경우 사용한다.  
&nbsp;&nbsp;&nbsp;&nbsp;에러 발생 여부와 관계없이 실행할 문장  
&nbsp;&nbsp;&nbsp;&nbsp;...

```python
try:
    print(4 / 1)
    print(4 / 0) # ZeroDivisionError 발생
    print(4 / 4) # 무시
except: # except 뒤에 에러 이름을 생략하면 모든 에러를 처리할 수 있다.
    # 에러 발생시 에러를 무시하려면 pass를 사용하다.
    # pass
    print('나눗셈은 0으로 할 수 없습니다.')
print('꺄~~~~~')
```

```python
data = [1, 2, 3]
try:
    print(data[2])
    print(data[3]) # IndexError 발생
    print(data[-1]) # 무시
except:
    print('리스트의 인덱스 지정이 올바르지 않습니다.')
print('꺄22~~~~~')
```

```python
data = [1, 2, 3]
string = '안녕~~~~~'
try:
    # print(4 / 0) # ZeroDivisionError 발생
    # print(data[3]) # IndexError 발생
    print(string)
    # print('{:d}'.format(string))
except ZeroDivisionError: # except 뒤에 에러 이름을 입력하면 지정된 에러만 처리한다.
    print('나눗셈은 0으로 할 수 없습니다.')
except IndexError as e: # except 뒤에 에러 이름을 쓰고 as 다음에 변수를 지정하면 에러 메시지가 변수에 저장된다.
    print('리스트의 인덱스 지정이 올바르지 않습니다.')
    print(e)
# except 뒤에 에러 이름을 지정하지 않거나 Exception를 입력하면 모든 에러를 처리하기 때문에 반드시 맨 마지막에 코딩해야 한다.
# ZeroDivisionError, IndexError를 제외한 모든 에러는 이곳에 코딩된 문장을 실행한다.
except Exception as e:
    print('데이터 타입이 올바르지 않습니다.')
    print(e)
# 에러 발생 여부와 관계없이 무조건 실행해야 할 문장이 있다면 사용한다.
finally:
    print('에러 발생 여부와 관계없이 무조건 실행해야 할 문장이 있다면 사용한다.')
print('꺄333~~~~~')
```
