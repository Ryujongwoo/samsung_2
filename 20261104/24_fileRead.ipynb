텍스트 파일을 읽기용으로 열 때는 디스크에 열려고하는 파일이 없을 수 있으므로 try ~ except를 사용해서 에러 처리를 한다.

## 한 글자씩 읽기

```python
try:
    file = open('.\\data\\data3.txt', 'r', encoding='utf-8')
    # 텍스트 파일에 저장된 데이터가 몇 줄인지 알 수 없으므로 무한 루프로 처리한다.
    while True:
        # read() 메소드는 텍스트 파일에 저장된 데이터를 문자 단위로 읽는다.
        char = file.read()
        # 텍스트 파일에서 읽어들인 문자가 없으면 무한 루프를 탈출한다.
        if char == '':
            break
        # ===== if
        print(char)
    # ===== while
    file.close()
except Exception as e:
    print('data3.txt 파일이 존재하지 않습니다.')
    print(e)
```

## 한 줄씩 읽기

```python
try:
    file = open('.\\data\\data3.txt', 'r', encoding='utf-8')
    
    # 텍스트 파일의 데이터를 읽어서 저장할 빈 리스트를 선언한다.
    lines = []
    
    while True:
        # readline() 메소드는 텍스트 파일에서 한 줄씩 읽어온다.
        line = file.readline()
        # 텍스트 파일에서 읽어들인 줄이 없으면 무한 루프를 탈출한다.
        if not line:
            break
        # print(line, end='')
        # print(line.strip())
        line = line.strip()
        if len(line) != 0:
            lines.append(line.strip())
except:
    print('data3.txt 파일이 존재하지 않습니다.')
    
print(lines)
```

## 텍스트 파일의 내용 전체를 한꺼번에 읽기

```python
try:
    file = open('.\\data\\data3.txt', 'r', encoding='utf-8')
    # readlines() 메소드는 텍스트 파일 전체를 한꺼번에 읽어서 각 줄 단위로 리스트를 만들어 리턴한다.
    lines = file.readlines()
    print(type(lines))
except:
    print('data3.txt 파일이 존재하지 않습니다.')
    
print(lines)
```
