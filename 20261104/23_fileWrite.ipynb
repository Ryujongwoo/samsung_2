# 텍스트 파일로 출력하기

파이썬을 이용해서 텍스트 파일을 처리할 경우 파일 탐색기에서 아래의 설정을 확인한다.  
보기 메뉴 => 표시 => 파일 확장명 체크  
... => 옵션 => 보기 탭으로 이동 => 고급 설정에서 알려진 파일 형식을 ~~~~~ 체크 해제

텍스트 파일로 출력하거나 텍스트 파일에서 읽어오는 경우 반드시 open() 함수를 사용해서 파일을 열어야 한다.

## open() 함수

변수 = `open('파일 이름', '파일 열기 모드')`  

파일 열기 모드  
`w`: write, 쓰기(덮어쓰기), 지정한 파일이 없으면 파일을 새로 만든다. 파일이 있으면 기존 내용을 지우고 저장한다.  
`a`: append, 쓰기(추가), 지정한 파일이 없으면 파일을 새로 만든다. 파일이 있으면 기존 내용 뒤에 추가해서 저장한다.  
`r`: read, 읽기, 지정한 파일이 없으면 에러가 발생된다.

`w`나 `a` 모드로 파일을 열어서 작업했으면 반드시 close() 메소드를 실행해서 파일을 닫아야 정상적으로 저장된다.  
`r` 모드로 열어서 작업했으면 파일을 닫지 않아도 상관없다.

파일을 open할 때 한글이 출력되지 않거나 또는 한글이 깨지거나 또는 cp949라는 메시지가 보이며 에러가 발생될 때 `encoding`을 `utf-8`로 지정한다.

```python
file = open('data.txt', 'w', encoding='utf-8') # data.txt 파일을 출력(덮어쓰기)용으로 열어준다.

for i in range(10):
    # 화면에 출력할 때는 print() 함수를 사용하지만 파일로 출력할 때는 write() 메소드를 사용한다.
    # 출력할 때 끝에 개행문자(\n)를 붙여야 텍스트 파일로 저장할 때 줄이 바뀐다.
    file.write('꺄아~~~~~' + '\n')
file.close()
    
print('data.txt 파일에 쓰기 완료')
```

## 경로지정 방법

`절대경로` 지정방법과 `상대경로` 지정방법이 있다.  
절대경로는 디스크 드라이브의 최상위(root) 폴더(디렉토리)부터 파일이 저장된 폴더까지의 경로를 의미한다.  
상대경로는 현재 화면에 표시되는 소스 프로그램이 위치한 폴더부터 파일이 저장된 폴더까지의 경로를 의미한다.  

절대경로나 상대경로를 입력할 때 폴더와 폴더 또는 폴더와 파일을 구분하는 구분자로 `\`가 기본적으로 사용되는데 `\n`이나 `\t`와 같은 이스케이프 시퀀스 때문에 에러가 발생될 수 있으므로 `\`를 `\\`나 `/`로 수정한다. => `\\`를 사용하는 것을 권장한다.

`C:\\workspace\\data\\data.txt` => 절대경로  
`C:/workspace/data/data.txt` => 절대경로  

`.`은 현재 작업중인 소스 프로그램이 저장된 폴더를 의미하고 `..`은 현재 작업중인 소스 프로그램이 저장된 폴더의 한 단계 상위 폴더를 의미한다.

`.\\data\\data.txt` => 상대경로  
`./data/data.txt` => 상대경로

```python
file = open('.\\data\\data.txt', 'w', encoding='utf-8')
for i in range(10):
    file.write('꺄아~~~~~' + '\n')
file.close()
print('data.txt 파일에 쓰기 완료')
```

## 키보드로 입력한 문자열을 텍스트 파일로 저장한다.

quit가 입력되면 저장 후 종료한다. quit가 대문자로 입력되던 소문자로 입력되던 섞여서 입력되던 저장한다.

```python
file = open('.\\data\\data2.txt', 'w', encoding='utf-8')
while True:
    string = input('>>> ')
    if string.lower() == 'quit':
        break
    # ===== if
    file.write(string + '\n')
file.close()
print('data2.txt 파일에 쓰기 완료')
```

## with를 사용해서 파일을 open 하면 with 블록의 모든 내용이 실행된 후 파일을 자동으로 닫아준다.

```python
with open('.\\data\\data3.txt', 'w', encoding='utf-8') as file:
    while True:
        string = input('>>> ')
        if string.lower() == 'quit':
            break
        file.write(string + '\n')
    # file.close()
print('data3.txt 파일에 쓰기 완료')  
```

```python
with open('.\\data\\data3.txt', 'a', encoding='utf-8') as file:
    while True:
        string = input('>>> ')
        if string.lower() == 'quit':
            break
        file.write(string + '\n')
    # file.close()
print('data3.txt 파일에 추가 완료')  
```
