```python
import warnings
warnings.filterwarnings(action='ignore')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import koreanize_matplotlib
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

## 식습관 데이터를 차원 축소시켜 시각화하기

사람들의 하루 식습관 데이터를 만든 후 주성분분석을 사용해서 시각화하기 위해 칼로리, 아침, 점심, 저녁, 운동횟수, 체형이라는 특징을 가지는 데이터프레임을 만든다.

```python
df = pd.DataFrame(columns = ['calorie', 'breakfast', 'lunch', 'dinner', 'exercise', 'body_shape'])
df.loc[0] = [1200, 1, 0, 0, 2, 'Skinny']
df.loc[1] = [2800, 1, 1, 1, 1, 'Normal']
df.loc[2] = [3500, 2, 2, 1, 0, 'Fat']
df.loc[3] = [1400, 0, 1, 0, 3, 'Skinny']
df.loc[4] = [5000, 2, 2, 2, 0, 'Fat']
df.loc[5] = [1300, 0, 0, 1, 2, 'Skinny']
df.loc[6] = [3000, 1, 0, 1, 1, 'Normal']
df.loc[7] = [4000, 2, 2, 2, 0, 'Fat']
df.loc[8] = [2600, 0, 2, 0, 0, 'Normal']
df.loc[9] = [3000, 1, 2, 1, 1, 'Fat']
df
```

데이터 전처리 과정을 통해서 데이터 범위가 제각각인 특징을 표준화한다.

위의 데이터는 칼로리가 유난히 다른 피처들에 비해 수치 범위가 큰 것을 확인할 수 있다.  
이 경우 모든 피처들이 평균이 0이고 분산이 1인 데이터로 변환한 후 비교하면 특성들의 상관 관계를 이해하기 쉬워지고 머신러닝 모델 역시 더욱 효율적으로 데이터를 학습할 수 있다.

피처 분리하기

```python
# 데이터에 피처와 레이블이 함께있으므로 피처로만 구성된 데이터프레임을 만든다.
x = df[['calorie', 'breakfast', 'lunch', 'dinner', 'exercise']]
x
```

칼로리를 포함한 모든 특성들을 같은 범위로 표준화한다.

```python
x_train = StandardScaler().fit_transform(x)
print(x_train)
```

레이블 분리하기

```python
y_train = df[['body_shape']]
y_train
```

공분산 행렬 구하기

주성분분석을 하기위해서 가장 먼저 할 일을 피처들의 공분산 행렬을 구하는 것이다.

```python
cov_mat = np.cov(x_train.T)
print(cov_mat)
```

고유값과 고유 벡터 구하기

공분산 행렬에서 가장 큰 고유값을 가진 벡터를 찾아 그 벡터위에 사영시킨다.

```python
las, evs = np.linalg.eig(cov_mat)
print('고유값\n', las, sep='')
print('고유 벡터\n', evs, sep='')
```

```python
# 각각의 고유 벡터로 데이터를 사영할 경우 얼마 만큼의 정보가 유지되는지 확인한다.
for i in range(len(las)):
    print(las[i] / sum(las))
```

위의 결과를 통해 1차원 데이터로 차원 축소를 하더라도 73.2% 정도에 해당되는 정보를 유지할 수 있다는 사실을 확인할 수 있고 26.8% 정도의 정보 손실은 감수하고 1차원 데이터로 축소한다.

evs.T[0]는 원래 첫 번째 고유 벡터를 얻어온다.  
np.linalg.norm(evs.T[0])는 첫 번째 고유 벡터를 자신의 크기로 나누어 길이가 1인 단위 벡터를 만든다.

```python
proj = x_train.dot(evs.T[0] / np.linalg.norm(evs.T[0]))
print(proj)
```

시각화

데이터가 사용된 고유 벡터(주성분)를 x축으로 하고 1차원 데이터이므로 y축은 0으로 고정시킨다.

```python
result = pd.DataFrame(proj, columns=['x'])
result['y'] = 0.0
result['label'] = y_train
result
```

```python
plt.figure(figsize=[6, 6])
sns.lmplot(data=result, x='x', y='y', hue='label', fit_reg=False, scatter_kws={'s': 20})
plt.title('차원 축소 결과', pad=30, fontsize=20)
plt.show()
```

# 사이킷런을 이용한 주성분 분석 구현

사이킷런 라이브러리에서 주성분분석을 사용하기 위해 PCA를 import 한다.

```python
from sklearn.decomposition import PCA
```

PCA 모델을 만들 때 n_components에 축소하려는 차원의 크기를 지정한다.

```python
model = PCA(n_components=1).fit_transform(x_train)
model
```

```python
result = pd.DataFrame(model, columns=['x'])
result['y'] = 0.0
result['label'] = y_train
result
```

```python
plt.figure(figsize=[6, 6])
sns.lmplot(data=result, x='x', y='y', hue='label', fit_reg=False, scatter_kws={'s': 20})
plt.title('차원 축소 결과', pad=30, fontsize=20)
plt.show()
```
