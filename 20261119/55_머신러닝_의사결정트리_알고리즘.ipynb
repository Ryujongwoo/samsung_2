# 의사결정 트리(Decision Tree) 알고리즘

데이터 분류나 회귀에 사용하는 지도학습 알고리즘으로 결과값이 어떻게 나왔는지 이해하기가 가장 `쉽고 정확도가 높다는 장점`이 있지만 `과적합되기 쉬운` 알고리즘이라는 단점도 있다.

의사결정 트리 알고리즘은 스므고개와 상당히 비슷하다. 데이터의 특징을 바탕으로 데이터를 연속으로 분리하다 보면 결국 하나의 정답으로 데이터를 분류할 수 있는데 스므고개에서 적은수의 질문으로 정답을 맞추기 위해서는 의미있는 질문을 먼저하는 것이 중요하듯이 의사결정 트리에서도 `의미있는 질문`을 먼저하는 것이 상당히 중요하다. 의사결정 트리는 테스트를 위쪽에 놓고 해당 테스트로 발생 가능한 결과를 아래쪽으로 놓는다. 상위 테스트로 판별하지 못하는 데이터는 또 다른 테스트인 하위 테스트를 사용해서 분류가 가능하다.

의사결정 트리는 테스트 성능을 평가하기 위해 `엔트로피(entropy)`라는 개념을 사용한다.  
엔트로피는 분순도 정도를 측정하며, 낮을수록 좋다. 이때, 불순도는 노드에 서로 다른 데이터가 얼마나 섞여있는지 확인한다. 불순도가 낮을수록 데이터가 섞여있지 않다는 의미이다.

## 지니 계수(Gini Index)

지니 계수는 엔트로피와 함께 불순도를 측정하는 또 다른 방법으로 P<sup>2</sup> + Q<sup>2</sup>으로 계산하며 데이터의 통계적 분석 정도를 정량화해서 표현하는 것을 의미한다. `사이킷런의 의사결정 트리는 트리를 구성할 때 지니 계수를 이용`한다.

## 와인 데이터를 의사결정 트리 알고리즘에 사용할 데이터를 준비한다.

```python
raw_data = datasets.load_wine()

xData = raw_data.data
yData = raw_data.target

x_train, x_test, y_train, y_test = train_test_split(xData, yData, random_state=0)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

## 의사결정 트리 모델을 만들고 학습 시킨다.

의사결정 트리 알고리즘을 사용하기 위해 DecisionTreeClassifier를 import 한다.

```python
from sklearn.tree import DecisionTreeClassifier
```

사이킷런의 의사결정 트리 알고리즘은 구동시 random 하게 특성들과 데이터 건수를 선택하는 로직이 포함되어 있어서 random_state에 특정한 값을 지정해서 수행시마다 동일한 트리를 만들 수 있다.

```python
model = DecisionTreeClassifier(random_state=0)
model.fit(x_train, y_train)
```

## 학습된 의사결정 트리 모델을 테스트 데이터로 예측한다.

```python
predict = model.predict(x_test)
print(predict)
```

```python
predict_proba = model.predict_proba(x_test)
print(predict_proba)
```

## 모델을 평가한다.

```python
classification = classification_report(y_test, predict, target_names=raw_data.target_names)
print(classification)
```
